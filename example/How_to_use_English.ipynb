```python
# Import required modules
import os
import numpy as np
import matplotlib.pyplot as plt
from pyplogo.extractors.structure_based import StructureExtractor
from pyplogo.visualizers.secondary_structure import SecondaryStructureVisualizer
from pyplogo.visualizers.themes import get_theme, get_all_themes
from pyplogo.utils.parsers import parse_fasta, validate_sequence
from pyplogo.utils.formatters import format_sequence, format_secondary_structure
from pyplogo.data.aa_properties import AA_PROPERTIES, AA_COLORS, SS_COLORS

# Set matplotlib font support (if needed)
plt.rcParams['font.family'] = ['DejaVu Sans', 'Arial']
plt.rcParams['axes.unicode_minus'] = False
```

```python
# Initialize extractor
extractor = StructureExtractor()

# Example PDB file path (needs to be replaced with actual file path)
pdb_file = r"example.pdb"  # Please replace with actual file path
chain_id = "A"  # Specify the chain to extract

try:
    # Extract structure data from PDB file
    structure_data = extractor.from_pdb(pdb_file, chain_id)
    
    print(f"Extracted sequence length: {len(structure_data.sequence)}")
    print(f"Secondary structure length: {len(structure_data.secondary_structure)}")
    print(f"Number of disulfide bonds: {len(structure_data.disulfide_bonds)}")
    
    # Display first 50 amino acids and corresponding secondary structure
    print("\nSequence (first 50):")
    print(structure_data.sequence[:50])
    
    print("\nSecondary structure (first 50):")
    print(structure_data.secondary_structure[:50])
    
    # Initialize visualizer
    visualizer = SecondaryStructureVisualizer(
        row_length=50,  # Display 50 amino acids per row
        amino_acid_spacing=28,  # Amino acid spacing
        aa_font_size=12,  # Amino acid font size
    )
    
    # Create visualization figure
    fig = visualizer.create_figure(
        sequence=structure_data.sequence,
        secondary_structure=structure_data.secondary_structure,
        title="Protein Secondary Structure Visualization",
        show_residue_numbers=True,  # Show residue numbers
        show_disulfide_bonds=True,  # Show disulfide bonds
        disulfide_bonds=structure_data.disulfide_bonds,  # Disulfide bond data
        figsize=(12, 8),  # Figure size
    )
    
    # Display figure
    plt.tight_layout()
    plt.show()
    
except Exception as e:
    print(f"Error: {e}")
    print("Please ensure a valid PDB file path is provided")
```

```python
from Bio import SeqIO

# Read FASTA file
fasta_file = "example.fasta"  # Replace with your FASTA file path

try:
    # Read FASTA file using Biopython
    records = list(SeqIO.parse(fasta_file, "fasta"))
    if not records:
        raise ValueError("FASTA file is empty or no valid sequence found")
    
    # Get first sequence
    record = records[0]
    sequence_from_fasta = str(record.seq)
    
    print(f"Sequence length parsed from FASTA: {len(sequence_from_fasta)}")
    print(f"Sequence ID: {record.id}")
    print(f"Description: {record.description}")
    print(f"First 50 amino acids of sequence: {sequence_from_fasta[:50]}")
    
    # Validate sequence
    is_valid = validate_sequence(sequence_from_fasta)
    print(f"Sequence validity: {is_valid}")
    
    # Use parsed sequence for visualization (assuming secondary structure)
    # In practical applications, secondary structure may need to be obtained from other sources
    hypothetical_ss = "H" * 20 + "C" * 10 + "T" * 5 + "E" * 15 + "C" * 10
    
    visualizer = SecondaryStructureVisualizer(
        row_length=50,
        aa_font_size=12,
    )
    
    fig = visualizer.create_figure(
        sequence=sequence_from_fasta,
        secondary_structure=hypothetical_ss,
        title=f"{record.description} Secondary Structure Visualization",
        show_residue_numbers=True,
        figsize=(20, 10),  # Figure size
        legend_position='right',  # Legend on the right
        ss_range=(1, len(hypothetical_ss)),  # Assume secondary structure covers entire sequence
        fill_gaps=False,  # New parameter: control whether uncovered regions are displayed as coil
    )
    
    plt.tight_layout()
    plt.show()
    
except Exception as e:
    print(f"Error: {e}")
```

```python
# Assume predicted secondary structure
predicted_ss = "CCHHHHHHHHHHHHHHHHHHHHHHHHCCCTTSCCSSEEEEEEEEEEESEEEEEEEEECSCHHHHHHHHHHHHHHHHHTSCHHHHHHHHHHHHHHHCCCCCTTTHHHHHHHHHHHHHTSHHHHTCCCSEEEEEEEEEECETTECEEEEEEECSSHHHHHHHHHHHHHHHHHGGGTSCCCCCCCC"

# Experimental structure
visualizer1 = SecondaryStructureVisualizer(
    row_length=50,
    aa_font_size=10,
)
print(structure_data.secondary_structure)
visualizer1.create_figure(
    sequence=structure_data.sequence,
    secondary_structure=structure_data.secondary_structure,
    title="Experimental Structure",
    show_residue_numbers=False,
    show_legend=True,
    show_disulfide_bonds=True,
    disulfide_bonds=structure_data.disulfide_bonds,
    ss_range=(1, len(structure_data.secondary_structure)),
    figsize=(12, 8)
)

# Predicted structure
visualizer2 = SecondaryStructureVisualizer(
    row_length=50,
    aa_font_size=10,
    aa_font_family=['DejaVu Sans', 'SimHei'] # Use fonts that support Chinese
)

visualizer2.create_figure(
    sequence=structure_data.sequence,
    secondary_structure=predicted_ss,
    title="Predicted Structure",
    show_residue_numbers=False,
    show_legend=False,
    ss_range=(1, len(predicted_ss)),
    figsize=(12, 8)
)

plt.tight_layout()
plt.show()

# Calculate agreement
def calculate_agreement(exp_ss, pred_ss):
    if len(exp_ss) != len(pred_ss):
        return 0
    
    agreement = sum(1 for e, p in zip(exp_ss, pred_ss) if e == p)
    return agreement / len(exp_ss)

agreement_score = calculate_agreement(structure_data.secondary_structure, predicted_ss)
print(f"Agreement between experimental and predicted secondary structure: {agreement_score:.2%}")
```
